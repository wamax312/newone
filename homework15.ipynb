```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests

%matplotlib inline
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

def generate_gpt_insight(prompt, api_key, timeout=60):
    url = "https://api.openai.com/v1/engines/text-davinci-003/completions"
    headers = {"Authorization": f"Bearer {api_key}"}
    data = {"prompt": prompt, "max_tokens": 150}
    try:
        response = requests.post(url, headers=headers, json=data, timeout=timeout)
        response.raise_for_status()
        return response.json()['choices'][0]['text'].strip()
    except requests.exceptions.RequestException as e:
        print(f"GPT API 请求失败: {e}")
        return None

YOUR_API_KEY = 'YOUR_API_KEY_HERE'

csv_file_path = r"D:\常用软件\vscode\daolun\users_combined_info_500.csv"

try:
    df = pd.read_csv(csv_file_path)
except FileNotFoundError:
    print(f"错误：找不到文件，请检查路径：{csv_file_path}")
    exit()
except Exception as e:
    print(f"读取文件时发生错误：{e}")
    exit()

df = df.drop(['name'], axis=1)

print("\n数据前几行：")
print(df.head())

print("\n数据类型信息：")
print(df.dtypes)

print("\n缺失值检查：")
print(df.isnull().sum())

df['event_time'] = pd.to_datetime(df['event_time'], errors='coerce')

print("\nevent_time列样本：")
print(df['event_time'].head())

print("\n---人口统计分析---")

country_counts = df['country'].value_counts(dropna=True)
if not country_counts.empty:
    plt.figure(figsize=(8, 6))
    top_countries = country_counts.head(10)
    sns.barplot(x=top_countries.values, y=top_countries.index, palette="viridis")
    plt.title('用户国家分布（前十）')
    plt.xlabel('用户数量')
    plt.ylabel('国家')
    plt.tight_layout()
    plt.show()
    
    country_distribution_insight = generate_gpt_insight("请解释这张关于GitHub用户所在国家分布的图表。", YOUR_API_KEY)
    if country_distribution_insight:
        print(country_distribution_insight)
else:
    print("国家分布数据为空，无法绘制图表。")

location_counts = df['location'].value_counts(dropna=True)
if not location_counts.empty:
    plt.figure(figsize=(8, 6))
    top_locations = location_counts.head(10)
    sns.barplot(x=top_locations.values, y=top_locations.index, palette="mako")
    plt.title('用户城市分布（前十）')
    plt.xlabel('用户数量')
    plt.ylabel('城市')
    plt.tight_layout()
    plt.show()
    
    city_distribution_insight = generate_gpt_insight("请解释这张关于GitHub用户所在城市的图表。", YOUR_API_KEY)
    if city_distribution_insight:
        print(city_distribution_insight)
else:
    print("城市分布数据为空，无法绘制图表。")

time_zone_counts = df['event_time'].dt.hour.value_counts().sort_index()
if not time_zone_counts.empty:
    plt.figure(figsize=(8, 6))
    sns.barplot(x=time_zone_counts.index, y=time_zone_counts.values, palette="rocket")
    plt.title('用户活动时间分布（小时）')
    plt.xlabel('小时')
    plt.ylabel('事件数量')
    plt.tight_layout()
    plt.show()
    
    timezone_distribution_insight = generate_gpt_insight("请解释这张关于GitHub用户活动时间分布的图表。", YOUR_API_KEY)
    if timezone_distribution_insight:
        print(timezone_distribution_insight)
else:
    print("时区分布数据为空，无法绘制图表。")

print("\n---协作行为分析---")

user_submission_counts = df['user_id'].value_counts()
if not user_submission_counts.empty:
    plt.figure(figsize=(8, 6))
    sns.histplot(user_submission_counts.values, bins=30, kde=True, color="skyblue")
    plt.title("提交频率分布")
    plt.xlabel("提交次数")
    plt.ylabel("用户数量")
    plt.tight_layout()
    plt.show()
    
    submission_frequency_insight = generate_gpt_insight("请解释这张关于GitHub用户提交频率分布的图表。", YOUR_API_KEY)
    if submission_frequency_insight:
        print(submission_frequency_insight)
else:
    print("提交频率数据为空，无法绘制图表。")

print("\n---其他有趣的洞察---")

event_type_counts = df['event_type'].value_counts()
if not event_type_counts.empty:
    plt.figure(figsize=(8, 6))
    sns.barplot(x=event_type_counts.values, y=event_type_counts.index, palette="flare")
    plt.title('事件类型分布')
    plt.xlabel('事件数量')
    plt.ylabel('事件类型')
    plt.tight_layout()
    plt.show()
    
    event_type_insight = generate_gpt_insight("请解释这张关于GitHub用户事件类型分布的图表。", YOUR_API_KEY)
    if event_type_insight:
        print(event_type_insight)
else:
    print("事件类型分布数据为空，无法绘制图表。")

df['event_hour'] = df['event_time'].dt.hour
event_time_distribution = df.groupby(['event_type', 'event_hour']).size().unstack(fill_value=0)

if event_time_distribution.empty:
    print("警告：event_time_distribution 为空，跳过热图绘制。")
else:
    event_time_distribution = event_time_distribution.loc[:, (event_time_distribution != 0).any(axis=0)]
    
    if event_time_distribution.empty:
        print("警告：event_time_distribution所有列均为0，跳过热图绘制。")
    else:
        plt.figure(figsize=(12, 8))
        sns.heatmap(event_time_distribution, annot=True, fmt="d", cmap="YlGnBu")
        plt.title("事件类型与时间分布")
        plt.xlabel('小时')
        plt.ylabel('事件类型')
        plt.tight_layout()
        plt.show()
        
        event_time_heatmap_insight = generate_gpt_insight("请解释这张关于GitHub用户事件类型与时间分布的热图。", YOUR_API_KEY)
        if event_time_heatmap_insight:
            print(event_time_heatmap_insight)

print("\n---更多有趣的洞察---")

df['event_date'] = df['event_time'].dt.date
event_type_trend = df.groupby(['event_date', 'event_type']).size().unstack(fill_value=0)

if not event_type_trend.empty:
    plt.figure(figsize=(12, 6))
    event_type_trend.plot(kind='line')
    plt.title('不同事件类型随时间的变化趋势')
    plt.xlabel('日期')
    plt.ylabel('事件数量')
    plt.tight_layout()
    plt.show()
    
    event_type_trend_insight = generate_gpt_insight("请解释这张关于GitHub用户不同事件类型随时间变化的趋势图。", YOUR_API_KEY)
    if event_type_trend_insight:
        print(event_type_trend_insight)
else:
    print("事件类型时间趋势数据为空，无法绘制图表。")

if not df.empty and 'total_influence' in df.columns:
    plt.figure(figsize=(12, 6))
    sns.scatterplot(data=df, x='total_influence', y='location', hue='country', palette='viridis')
    plt.title('用户的地理位置与影响力关系')
    plt.xlabel('总影响力')
    plt.ylabel('地理位置')
    plt.tight_layout()
    plt.legend(title='国家')
    plt.show()
    
    location_influence_insight = generate_gpt_insight("请解释这张关于GitHub用户地理位置与影响力关系的散点图。", YOUR_API_KEY)
    if location_influence_insight:
        print(location_influence_insight)
else:
    print("地理位置与影响力关系数据为空，无法绘制图表。")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.backends.backend_pdf import PdfPages

# 确保在Jupyter Notebook中可以显示图表，并设置中文字体以正确显示汉字
%matplotlib inline
plt.rcParams['font.sans-serif'] = ['SimHei']  # 设置字体为黑体，确保中文显示正常
plt.rcParams['axes.unicode_minus'] = False    # 解决负号显示问题

# 读取CSV文件
csv_file_path = r"D:\常用软件\vscode\daolun\users_combined_info_500.csv"

try:
    df = pd.read_csv(csv_file_path)
except FileNotFoundError:
    print(f"错误：找不到文件，请检查路径：{csv_file_path}")
    exit()
except Exception as e:
    print(f"读取文件时发生错误：{e}")
    exit()

# 数据预处理与隐私保护：删除或匿名化敏感信息
df = df.drop(['name'], axis=1)  # 删除姓名列，假设没有邮箱列

# 初始化PDF文件
pdf_output = "data_insights_report.pdf"
with PdfPages(pdf_output) as pdf_pages:

    # 创建首页
    fig, ax = plt.subplots(figsize=(8.5, 11))  # A4纸大小
    ax.text(0.5, 0.5, "数据洞察报告\n基于GitHub用户活动数据分析", 
            fontsize=24, ha='center', va='center')
    ax.axis('off')  # 关闭坐标轴
    pdf_pages.savefig(fig)
    plt.close(fig)

    # ---人口统计分析---
    print("\n---人口统计分析---")

    # 国家和地区分布
    country_counts = df['country'].value_counts(dropna=True)  # 排除缺失值
    if not country_counts.empty:
        top_countries = country_counts.head(10)  # 只显示前10个国家
        fig, ax = plt.subplots(figsize=(8, 6))
        sns.barplot(x=top_countries.values, y=top_countries.index, palette="viridis", ax=ax)
        ax.set_title('用户国家分布（前十）\n这张图表展示了来自不同国家的GitHub用户的分布情况。\n我们可以看到美国用户数量最多，其次是其他几个主要科技中心国家。')
        ax.set_xlabel('用户数量')
        ax.set_ylabel('国家')
        plt.tight_layout()  # 调整布局以防止标签被截断
        pdf_pages.savefig(fig)
        plt.close(fig)

    # 城市级别分布
    location_counts = df['location'].value_counts(dropna=True)
    if not location_counts.empty:
        top_locations = location_counts.head(10)  # 只显示前10个地点
        fig, ax = plt.subplots(figsize=(8, 6))
        sns.barplot(x=top_locations.values, y=top_locations.index, palette="mako", ax=ax)
        ax.set_title('用户城市分布（前十）\n此图表示了GitHub用户最集中的前十个城市。\n可以看出，科技中心如旧金山湾区的城市有较多用户。')
        ax.set_xlabel('用户数量')
        ax.set_ylabel('城市')
        plt.tight_layout()
        pdf_pages.savefig(fig)
        plt.close(fig)

    # 时区分布
    if 'event_time' in df.columns:
        try:
            df['event_time'] = pd.to_datetime(df['event_time'], errors='coerce')
            time_zone_counts = df['event_time'].dt.hour.value_counts().sort_index()
            if not time_zone_counts.empty:
                fig, ax = plt.subplots(figsize=(8, 6))
                sns.barplot(x=time_zone_counts.index, y=time_zone_counts.values, palette="rocket", ax=ax)
                ax.set_title('用户活动时间分布（小时）\n这个图表显示了GitHub用户活跃的时间分布。\n可以看到，在工作时间（如上午9点至下午5点）用户活动最为频繁。')
                ax.set_xlabel('小时')
                ax.set_ylabel('事件数量')
                plt.tight_layout()
                pdf_pages.savefig(fig)
                plt.close(fig)
        except Exception as e:
            print(f"处理event_time列时发生错误: {e}")

    # ---协作行为分析---
    print("\n---协作行为分析---")

    # 提交频率
    user_submission_counts = df['user_id'].value_counts()
    if not user_submission_counts.empty:
        fig, ax = plt.subplots(figsize=(8, 6))
        sns.histplot(user_submission_counts.values, bins=30, kde=True, color="skyblue", ax=ax)
        ax.set_title("提交频率分布\n这张图展示了每个用户提交的频率。\n大多数用户提交次数较少，但也有部分用户非常活跃。")
        ax.set_xlabel("提交次数")
        ax.set_ylabel("用户数量")
        plt.tight_layout()
        pdf_pages.savefig(fig)
        plt.close(fig)

    # ---其他有趣的洞察---
    print("\n---其他有趣的洞察---")

    # 事件类型频率
    event_type_counts = df['event_type'].value_counts()
    if not event_type_counts.empty:
        fig, ax = plt.subplots(figsize=(8, 6))
        sns.barplot(x=event_type_counts.values, y=event_type_counts.index, palette="flare", ax=ax)
        ax.set_title('事件类型分布\n该图表展示了不同类型的GitHub事件发生的频率。\nCreateEvent是最常见的事件类型之一。')
        ax.set_xlabel('事件数量')
        ax.set_ylabel('事件类型')
        plt.tight_layout()
        pdf_pages.savefig(fig)
        plt.close(fig)

    # 不同类型的事件发生在哪个时间比较多
    if 'event_hour' in df.columns:
        event_time_distribution = df.groupby(['event_type', 'event_hour']).size().unstack(fill_value=0)

        if not event_time_distribution.empty:
            event_time_distribution = event_time_distribution.loc[:, (event_time_distribution != 0).any(axis=0)]
            if not event_time_distribution.empty:
                fig, ax = plt.subplots(figsize=(12, 8))
                sns.heatmap(event_time_distribution, annot=True, fmt="d", cmap="YlGnBu", ax=ax)
                ax.set_title("事件类型与时间分布\n热图显示了不同事件类型在一天内各个时间段的发生频率。\n例如，CreateEvent通常在工作时间内更常见。")
                ax.set_xlabel('小时')
                ax.set_ylabel('事件类型')
                plt.tight_layout()
                pdf_pages.savefig(fig)
                plt.close(fig)

    # 事件类型的时间趋势
    if 'event_date' in df.columns:
        event_type_trend = df.groupby(['event_date', 'event_type']).size().unstack(fill_value=0)

        if not event_type_trend.empty:
            fig, ax = plt.subplots(figsize=(12, 6))
            event_type_trend.plot(kind='line', ax=ax)
            ax.set_title('不同事件类型随时间的变化趋势\n这条线形图展示了不同类型事件随时间的变化趋势。\n我们可以观察到某些事件类型在特定时间段内的波动。')
            ax.set_xlabel('日期')
            ax.set_ylabel('事件数量')
            plt.tight_layout()
            pdf_pages.savefig(fig)
            plt.close(fig)

    # 地理位置与影响力的关系
    if not df.empty and 'total_influence' in df.columns and 'location' in df.columns:
        fig, ax = plt.subplots(figsize=(12, 6))
        sns.scatterplot(data=df, x='total_influence', y='location', hue='country', palette='viridis', ax=ax)
        ax.set_title('用户的地理位置与影响力关系\n散点图展示了用户的地理位置与其影响力之间的关系。\n不同的颜色代表不同的国家。')
        ax.set_xlabel('总影响力')
        ax.set_ylabel('地理位置')
        plt.tight_layout()
        pdf_pages.savefig(fig)
        plt.close(fig)

print(f"报告已保存至: {pdf_output}")
```
